```python
#Feature Scaling
# x- mean/ sigma
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_train_scaled = scaler.fit_transform(X_train) #Learn the scaling statistics and apply them.  Learn + scale
X_test_scaled = scaler.transform(X_test) #Reuse the training statistics without learning anything new from the test set. Scale Only
```

```python
import numpy as np
w=np.zeros(12)
b=0
```

```python
def Cost(X,y,w,b):
    summation=0
    m,n=X.shape
    for i in range(0,m):
        z = np.dot(w, X[i]) + b
        g_z=1/(1+np.exp(-z))
        summation+=y.iloc[i]*np.log(g_z) + (1-y.iloc[i])* np.log(1-g_z)
    J_cost=(-1/m)*summation
    return J_cost
```

```python
Cost(X_train_scaled,y_train,np.zeros(12),0)
```

```python
def differentiate(X,y,w,b):
    summation_w=0
    summation_b=0
    dj_dw=0
    dj_db=0
    m,n=X.shape
    for i in range(m):
        z=np.dot(w,X[i])+b
        g_z=1/(1+np.exp(-z))
        summation_w+=(g_z-y.iloc[i])*X[i]
        summation_b+=(g_z-y.iloc[i])
    dj_dw=(1/m)*summation_w
    dj_db=(1/m)*summation_b
    return dj_dw,dj_db
```

```python
differentiate(X_train_scaled, y_train,w,b)
```

```python
print("X shape:", X_train_scaled.shape)
print("y shape:", y.shape)
print("y values:", np.unique(y))

print("Mean of y:", np.mean(y))
print("Expected bias gradient:", 0.5 - np.mean(y))
```

```python
def GD(X,y,w,b,alpha):
    dj_dw, dj_db = differentiate(X, y, w, b)
    w=w-alpha*dj_dw
    b=b-alpha*dj_db

    return w,b
```

```python
GD(X_train_scaled, y_train, w, b, 0.01)
```

```python
def main(X,y):
    m,n=X.shape
    w=np.zeros(12)
    b=0
    alpha=0.01
    iterations=10000
    J_cost=Cost(X,y,w,b)
    for i in range(iterations):
        new_w,new_b=GD(X,y,w,b,alpha)
        w=new_w
        b=new_b
        new_cost=Cost(X,y,w,b)
        if(new_cost<J_cost):
            J_cost=new_cost
        else:
            return f'Gradient Descent is diverging.Try reducing the learning rate (alpha)'
    return J_cost,w,b
        
```

```python
main(X_train_scaled,y_train)
```
